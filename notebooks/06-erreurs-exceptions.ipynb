- un fournisseur sans livraison (`ZeroDivisionError`) : afficher `Aucune livraison pour ...`.

````python
delais = {
    "Acier SA": [3, 5, 4],
    "BoisPlus": [10, 12],
    "NouveauFournisseur": [],
}

def delai_moyen(fournisseur):
    # TODO : entoure ce calcul d'un try / except
    jours = delais[fournisseur]
    return sum(jours) / len(jours)

print(delai_moyen("Acier SA"))
print(delai_moyen("NouveauFournisseur"))
print(delai_moyen("Plastico"))

# Résultat attendu :
# 4.0
# Aucune livraison pour NouveauFournisseur
# None
# Fournisseur inconnu : Plastico
# None
````

<details>
<summary>💡 Solution</summary>

```python
def delai_moyen(fournisseur):
    try:
        jours = delais[fournisseur]
        return sum(jours) / len(jours)
    except KeyError:
        print(f"Fournisseur inconnu : {fournisseur}")
    except ZeroDivisionError:
        print(f"Aucune livraison pour {fournisseur}")
    return None

print(delai_moyen("Acier SA"))
print(delai_moyen("NouveauFournisseur"))
print(delai_moyen("Plastico"))
```

def commander(article, quantite):
    """Prépare une ligne de commande.

    Args:
        article: Nom de l'article.
        quantite: Nombre d'unités commandées.

    Returns:
        La ligne de commande sous forme de texte.

    Raises:
        TypeError: Si la quantité n'est pas un entier.
        ValueError: Si la quantité est inférieure ou égale à 0.
    """

        raise TypeError(f"La quantité doit être un entier, reçu {type(quantite).__name__}")

        raise ValueError(f"La quantité doit être positive, reçu {quantite}")
    return f"Commande : {quantite} x {article}"

print(commander("vis", 50))

for q in [0, "10", -2]:
    try:
        commander("vis", q)
    except (TypeError, ValueError) as e:
        print(f"{type(e).__name__} : {e}")

<details>
<summary>💡 Solution</summary>

```python
class EntrepotError(Exception):
    """Erreur de base pour la gestion d'entrepôt."""


class CapaciteDepasseeError(EntrepotError):
    """L'entrepôt n'a pas assez de place."""


def ranger(entrepot, palettes):
    places_libres = entrepot["capacite"] - entrepot["occupe"]
    if palettes > places_libres:
        raise CapaciteDepasseeError(
            f"{entrepot['nom']} : impossible de ranger {palettes} palettes, "
            f"il reste {places_libres} places"
        )
    entrepot["occupe"] += palettes


entrepot = {"nom": "Lyon", "capacite": 100, "occupe": 70}
for arrivage in [20, 15, 10]:
    try:
        ranger(entrepot, arrivage)
        print(f"{arrivage} palettes rangées ({entrepot['occupe']}/{entrepot['capacite']})")
    except CapaciteDepasseeError as e:
        print(f"⚠️ {e}")
```
</details>

🤔 Pourquoi créer `EntrepotError` alors qu'on ne l'utilise pas directement ? Parce qu'un `except EntrepotError` attrapera **toutes** les erreurs de ton application (capacité, référence inconnue, etc.) sans toucher aux vrais bugs Python.
