**Exercice 10**
```python
def ca_total(ventes):
    total = 0
    for v in ventes:
        total += v["montant"]
    return total

def filtrer_par_ville(ventes, ville):
    resultat = []
    for v in ventes:
        if v["ville"] == ville:
            resultat.append(v)
    return resultat

def rapport(ventes, ville):
    ventes_ville = filtrer_par_ville(ventes, ville)
    print(f"CA total : {ca_total(ventes):.2f} €")
    print(f"{ville} : {len(ventes_ville)} ventes, {ca_total(ventes_ville):.2f} €")
```
Remarque que `ca_total` sert deux fois : pour toutes les ventes, puis pour celles de la ville. C'est tout l'intérêt d'une fonction !
