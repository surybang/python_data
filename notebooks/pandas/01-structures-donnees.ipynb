## 8. `axis=0` ou `axis=1` ?

Voici les colis expédiés par trois entrepôts sur trois mois.

1. Calcule le **total de chaque mois** (une valeur par colonne).
2. Calcule le **total de chaque entrepôt** sur le trimestre (une valeur par ligne).

````python
expeditions = pd.DataFrame(
    {"janv": [100, 80, 60], "fev": [120, 70, 65], "mars": [90, 85, 70]},
    index=["Lille", "Lyon", "Nantes"],
)

# TODO

# Résultat attendu :
# janv    240
# fev     255
# mars    245
# dtype: int64
# Lille     310
# Lyon      235
# Nantes    195
# dtype: int64
````

<details>
<summary>💡 Solution</summary>

```python
print(expeditions.sum(axis=0))   # on « écrase » les lignes : une valeur par colonne
print(expeditions.sum(axis=1))   # on « écrase » les colonnes : une valeur par ligne
```
</details>

🤔 Ici, additionner les colonnes a un sens métier : ce sont trois mesures de même nature (des colis par mois).

## 10. Chasse au bug : la colonne qui disparaît

Ce code doit ajouter une colonne `poids_moyen` (poids par colis), puis l'afficher. Il plante avec une `KeyError`. Pourquoi ? Corrige-le.

````python
livraisons.assign(poids_moyen=livraisons["poids_kg"] / livraisons["colis"])
print(livraisons["poids_moyen"].round(2))

# Résultat attendu après correction :
# 0    2.54
# 1    1.60
# 2    1.78
# 3    3.05
# 4    1.50
# Name: poids_moyen, dtype: float64
````

<details>
<summary>💡 Solution</summary>

```python
# assign() ne modifie pas le DataFrame : il en RENVOIE un nouveau.
# Le résultat n'était stocké nulle part, la colonne n'existe donc pas dans livraisons.
livraisons = livraisons.assign(poids_moyen=livraisons["poids_kg"] / livraisons["colis"])
print(livraisons["poids_moyen"].round(2))
```
</details>

## 11. Pour aller plus loin : quel transporteur charge le plus ?

À partir de `livraisons`, affiche le **poids moyen par colis** le plus élevé, puis le **nom du transporteur** de cette livraison.

*Indice : `.max()` donne la valeur maximale, et `.idxmax()` donne l'étiquette de la ligne où elle se trouve.*

````python
# TODO

# Résultat attendu :
# 3.05
# Fretex
````

<details>
